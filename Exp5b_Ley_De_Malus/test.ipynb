```python
def numero_a_notacion_cientifica(numero, decimales=4):
    # Convierte el número a notación científica con el número de decimales especificado
    formato = f"{{:.{decimales}e}}"
    notacion_cientifica = formato.format(numero)
    
    # Divide la notación científica en la base y el exponente
    base, exponente = notacion_cientifica.split('e')
    exponente = int(exponente)
    
    # Genera la representación en LaTeX
    latex_resultado = f"{base} \\times 10^{{{exponente}}}"
    return latex_resultado

resultado_latex = numero_a_notacion_cientifica(0.002625, 3)
print(resultado_latex)
```
